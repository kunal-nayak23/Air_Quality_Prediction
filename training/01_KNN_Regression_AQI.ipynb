# K-Nearest Neighbors (KNN) Regression - AQI Prediction

## 1. Load and clean the data

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

data_path = '../data/raw/station_day.csv'
if not os.path.exists(data_path):
    data_path = 'data/raw/station_day.csv'
features = [
    'PM2.5', 'PM10', 'NO', 'NO2', 'NOx', 'NH3',
    'CO', 'SO2', 'O3', 'Benzene', 'Toluene'
]

data = pd.read_csv(data_path)
data = data.dropna(subset=['AQI', 'AQI_Bucket'])
data = data.drop_duplicates(subset=features + ['AQI', 'AQI_Bucket'])
data['Date'] = pd.to_datetime(data['Date'])
data = data.sort_values(['Date', 'StationId']).reset_index(drop=True)

print('Dataset shape:', data.shape)
data.head()
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
```

```python
print('Missing values in input columns:')
print(data[features].isnull().sum())
```

## 2. Split the data and create the model

```python
dates = data['Date'].drop_duplicates()
train_dates, test_dates = train_test_split(
    dates, test_size=0.2, random_state=42
)

train_data = data[data['Date'].isin(train_dates)]
test_data = data[data['Date'].isin(test_dates)]

X_train = train_data[features]
X_test = test_data[features]
y_train = train_data['AQI']
y_test = test_data['AQI']

print('Training data:', X_train.shape)
print('Testing data:', X_test.shape)
```

```python
model = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler()),
    ('model', KNeighborsRegressor(n_neighbors=7, weights='distance', n_jobs=4))
])
model
```

## 3. Train and test the model

```python
model.fit(X_train, y_train)
predictions = model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, predictions)

print('MAE:', round(mae, 4))
print('MSE:', round(mse, 4))
print('RMSE:', round(rmse, 4))
print('R2 Score:', round(r2, 4))
```

```python
sample = X_test.iloc[[0]]
actual_aqi = y_test.iloc[0]
predicted_aqi = model.predict(sample)[0]

print('Actual AQI:', round(actual_aqi, 2))
print('Predicted AQI:', round(predicted_aqi, 2))
```

```python
residuals = y_test - predictions

plt.figure(figsize=(7, 5))
plt.scatter(y_test, predictions, alpha=0.3)
plt.plot([0, y_test.max()], [0, y_test.max()], 'r--')
plt.xlabel('Actual AQI')
plt.ylabel('Predicted AQI')
plt.title('Actual vs Predicted AQI')
plt.show()

plt.figure(figsize=(7, 5))
plt.hist(residuals, bins=40)
plt.xlabel('Error')
plt.ylabel('Frequency')
plt.title('Error Distribution')
plt.show()
```

## 4. Compare different K values

```python
results = []
sample_rows = X_train.sample(n=15000, random_state=42)
sample_target = y_train.loc[sample_rows.index]

for k in [3, 5, 7, 11, 15, 21]:
    test_model = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', KNeighborsRegressor(n_neighbors=k, weights='distance', n_jobs=4))
    ])
    test_model.fit(sample_rows, sample_target)
    test_prediction = test_model.predict(X_test)
    test_rmse = np.sqrt(mean_squared_error(y_test, test_prediction))
    results.append([k, test_rmse])

k_results = pd.DataFrame(results, columns=['K', 'RMSE'])
k_results.plot(x='K', y='RMSE', marker='o')
plt.title('K Value vs RMSE')
plt.show()
```
